### 5.8 Defining Student Performance Groups

Students can have different levels of academic performance. To make our synthetic dataset more useful for analysis, we will define three performance groups: High, Average, and Needs Improvement.

We will use NumPy to randomly assign each student to one of these groups. Each group will have a different typical mark level.

These groups are used only to generate varied sample data. They will not determine the final grade or pass/fail result. Those results will be calculated later using the student's actual marks and the project's defined rules.

````python

# Set the number of students in our dataset
number_of_students = 500

# Randomly assign a performance group to each student
performance_groups = rng.choice(
    ["High", "Average", "Needs Improvement"],
    size=number_of_students,
    p=[0.25, 0.50, 0.25]
)

print("Total performance-group assignments:", len(performance_groups))
print("\nFirst 10 assignments:")
print(performance_groups[:10])

print("\nNumber of students in each group:")
print(pd.Series(performance_groups).value_counts())
````

### 5.9 Defining Typical Marks for Each Performance Group

Different performance groups will have different typical mark levels.

We will use a Python dictionary to store the typical mark level for each group. A dictionary stores information as key-value pairs.

For this dataset, we will use the following approximate mark levels:

- High: 84 marks
- Average: 66 marks
- Needs Improvement: 42 marks

NumPy will use these values when generating marks for individual subjects. The actual marks will vary between students and subjects.

These values are assumptions for synthetic data generation, not results from real student records.

````python

# Define the typical mark level for each performance group
group_means = {
    "High": 84,
    "Average": 66,
    "Needs Improvement": 42
}

print("Typical marks for each performance group:")

for group, mean_mark in group_means.items():
    print(group, ":", mean_mark)
````

### 5.10 Generating Subject Marks for One Student

We will use NumPy's normal distribution to generate marks around the typical mark level of a performance group.

The `loc` parameter specifies the center of the distribution, while `scale` controls how much the generated values vary. We will round the marks to whole numbers and keep them within the range of 0 to 100.

````python

# Select the first student's performance group
group = performance_groups[0]

# Find the typical mark level for that group
ability = group_means[group]

# Generate marks for five subjects
sample_student_marks = rng.normal(
    loc=ability,
    scale=11,
    size=len(subjects)
)

# Round marks and keep them between 0 and 100
sample_student_marks = np.clip(
    np.round(sample_student_marks),
    0,
    100
).astype(int)

print("Performance group:", group)
print("Typical mark level:", ability)
print("Subjects:", subjects)
print("Generated marks:", sample_student_marks)
````

### 5.12 Generating Student Records Using a Loop

A loop allows us to repeat the same operation multiple times. In this project, we will use a `for` loop to generate records for 500 students.

For every student, the program will assign a unique Student ID, a name, a department, marks in five subjects, and an attendance percentage.

Each student's information will be stored in a dictionary. The dictionaries will be collected in a Python list called `records`, which will later be converted into a Pandas DataFrame.

````python

# Create an empty list to store student records
records = []

# Generate records for all students
for i in range(number_of_students):

    # Get this student's performance group
    group = performance_groups[i]

    # Get the typical marks for that group
    ability = group_means[group]

    # Generate marks for the five subjects
    marks = rng.normal(
        loc=ability,
        scale=11,
        size=len(subjects)
    )

    # Round marks and restrict them to 0–100
    marks = np.clip(
        np.round(marks),
        0,
        100
    ).astype(int)

    # Create the student's complete record
    record = {
        "Student_ID": f"STU{i + 1:04d}",
        "Name": student_names[i],
        "Department": rng.choice(departments),
        "Python": marks[0],
        "Mathematics": marks[1],
        "AI": marks[2],
        "Statistics": marks[3],
        "Communication": marks[4],
        "Attendance": int(rng.integers(50, 101))
    }

    # Add the record to the list
    records.append(record)

print("Total student records generated:", len(records))
print("\nFirst student record:")
print(records[0])
````

### Part 4: Student Performance Analysis

In this section, we use **Pandas** to calculate each student's total and average marks across five subjects: Python, Mathematics, AI, Statistics, and Communication.

- **Total Marks:** The sum of marks obtained in all five subjects. The maximum possible total is 500.
- **Average Marks:** Total marks divided by the number of subjects. The maximum possible average is 100.

These calculations help us understand each student's overall academic performance.

````python

# Calculate total marks for each student
df["Total_Marks"] = df[subject_columns].sum(axis=1)

# Calculate average marks for each student
df["Average_Marks"] = df[subject_columns].mean(axis=1).round(2)

# Display the results for the first five students
display(
    df[["Student_ID", "Name", "Total_Marks", "Average_Marks"]].head()
)
````

### Part 5: Grading System and Pass/Fail Status

In this section, we assign grades to students based on their average marks. We also determine whether each student has passed or failed.

**Grading rules:**

| Average Marks | Grade |
|---|---|
| 90–100 | A+ |
| 80–89.99 | A |
| 70–79.99 | B |
| 60–69.99 | C |
| 50–59.99 | D |
| Below 50 | F |

**Pass/Fail rule:** A student passes if they score at least 40 marks in every subject. If they score below 40 in even one subject, they fail.

These rules help us classify students according to their academic performance.

````python

def calculate_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"


# Test the function with sample averages
print("Average 95:", calculate_grade(95))
print("Average 85:", calculate_grade(85))
print("Average 65:", calculate_grade(65))
print("Average 35:", calculate_grade(35))
````

````python

# Assign a grade to every student
df["Grade"] = df["Average_Marks"].apply(calculate_grade)

# Check whether each student passed all subjects
df["Result"] = np.where(
    df[subject_columns].ge(40).all(axis=1),
    "Pass",
    "Fail"
)

# Display student performance
display(
    df[
        [
            "Student_ID",
            "Name",
            "Average_Marks",
            "Grade",
            "Result"
        ]
    ].head(10)
)
````

### Part 6: Class Performance Summary

In this section, we analyze the overall academic results of the students.

We will calculate:
- Total number of students who passed.
- Total number of students who failed.
- Overall pass and fail percentages.
- Number of students in each grade category.

This analysis helps us understand the overall performance of the class.

````python

# Count students who passed and failed
pass_count = (df["Result"] == "Pass").sum()
fail_count = (df["Result"] == "Fail").sum()

# Calculate percentages
total_students = len(df)

pass_percentage = (pass_count / total_students) * 100
fail_percentage = (fail_count / total_students) * 100

# Display the summary
print("Total Students:", total_students)
print("Students Passed:", pass_count)
print("Students Failed:", fail_count)
print("Pass Percentage:", round(pass_percentage, 2), "%")
print("Fail Percentage:", round(fail_percentage, 2), "%")
````

````python

# Count the number of students in each grade
grade_distribution = df["Grade"].value_counts().reindex(
    ["A+", "A", "B", "C", "D", "F"],
    fill_value=0
)

print("Grade Distribution:")
display(grade_distribution.to_frame(name="Number_of_Students"))
````

### Part 7: Top Performers and Performance Ranking

In this section, we identify students based on their total and average marks.

We will:
- Find the student with the highest average marks.
- Find the student with the lowest average marks.
- Display the top 10 performing students.
- Rank students according to their average marks.

This analysis helps identify high achievers and students who may need additional academic support.

````python

# Find the index of the highest and lowest average marks
highest_index = df["Average_Marks"].idxmax()
lowest_index = df["Average_Marks"].idxmin()

# Get the corresponding student records
highest_performer = df.loc[highest_index]
lowest_performer = df.loc[lowest_index]

print("Highest Performer:")
display(
    highest_performer[
        ["Student_ID", "Name", "Department", "Average_Marks", "Grade"]
    ].to_frame().T
)

print("Lowest Performer:")
display(
    lowest_performer[
        ["Student_ID", "Name", "Department", "Average_Marks", "Grade"]
    ].to_frame().T
)
````

Display the top 10 students

````python

# Sort students from highest to lowest average marks
ranked_students = df.sort_values(
    by=["Average_Marks", "Student_ID"],
    ascending=[False, True]
).copy()

# Assign rankings
ranked_students["Rank"] = range(1, len(ranked_students) + 1)

# Display the top 10 performers
top_10_students = ranked_students[
    [
        "Rank",
        "Student_ID",
        "Name",
        "Department",
        "Total_Marks",
        "Average_Marks",
        "Grade",
        "Result"
    ]
].head(10)

display(top_10_students)
````

````python

main_code = r'''
from pathlib import Path
import numpy as np
import pandas as pd

PROJECT_DIR = Path(__file__).resolve().parent
DATA_FILE = PROJECT_DIR / "students.csv"
SUBJECTS = [
    "Python", "Mathematics", "AI",
    "Statistics", "Communication"
]


def calculate_grade(average):
    """Return a grade based on the student's average marks."""
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    return "F"


def calculate_student_performance(marks):
    """Calculate total and average marks."""
    marks = np.asarray(marks, dtype=float)
    return int(np.sum(marks)), round(float(np.mean(marks)), 2)


def analyze_students(df):
    """Calculate student results and create analysis reports."""
    df = df.copy()

    df["Total_Marks"] = df[SUBJECTS].sum(axis=1)
    df["Average_Marks"] = df[SUBJECTS].mean(axis=1).round(2)
    df["Grade"] = df["Average_Marks"].apply(calculate_grade)

    df["Result"] = np.where(
        df[SUBJECTS].ge(40).all(axis=1),
        "Pass",
        "Fail"
    )

    df["Rank"] = df["Average_Marks"].rank(
        method="first", ascending=False
    ).astype(int)

    return df


def main():
    if not DATA_FILE.exists():
        print("Error: students.csv was not found.")
        return

    df = pd.read_csv(DATA_FILE)

    missing_columns = [
        column for column in
        ["Student_ID", "Name", "Department", "Attendance"] + SUBJECTS
        if column not in df.columns
    ]

    if missing_columns:
        print("Missing required columns:", missing_columns)
        return

    df = analyze_students(df)

    output_file = PROJECT_DIR / "student_performance_report.csv"
    df.to_csv(output_file, index=False)

    subject_report = pd.DataFrame({
        "Average_Marks": df[SUBJECTS].mean().round(2),
        "Highest_Marks": df[SUBJECTS].max(),
        "Lowest_Marks": df[SUBJECTS].min()
    })
    subject_report.to_csv(
        PROJECT_DIR / "subject_performance_report.csv",
        index_label="Subject"
    )

    grade_counts = df["Grade"].value_counts().reindex(
        ["A+", "A", "B", "C", "D", "F"],
        fill_value=0
    )
    grade_counts.to_frame(
        name="Number_of_Students"
    ).to_csv(
        PROJECT_DIR / "grade_distribution_report.csv",
        index_label="Grade"
    )

    print("STUDENT PERFORMANCE ANALYTICS")
    print("-----------------------------")
    print("Total students:", len(df))
    print("Class average:", round(df["Average_Marks"].mean(), 2))
    print("Passed:", int((df["Result"] == "Pass").sum()))
    print("Failed:", int((df["Result"] == "Fail").sum()))

    print("\\nTop 10 students:")
    print(
        df.sort_values("Rank")[
            ["Rank", "Student_ID", "Name", "Average_Marks", "Grade", "Result"]
        ].head(10).to_string(index=False)
    )

    print("\\nSubject-wise analysis:")
    print(subject_report)

    print("\\nReports saved successfully.")


if __name__ == "__main__":
    main()
'''

main_path.write_text(main_code.strip() + "\n", encoding="utf-8")
print("main.py created successfully.")
print("File size:", main_path.stat().st_size, "bytes")
````
